## Берілген деректер

```python
states = ["A", "B", "C"]
T = [[0.7, 0.3, 0.0],      # A-дан
     [0.2, 0.6, 0.2],      # B-дан
     [0.0, 0.3, 0.7]]      # C-дан
emission = {"RED": [0.8, 0.5, 0.2], "BLUE": [0.2, 0.5, 0.8]}
b0 = [0.5, 0.3, 0.2]
readings = ["RED", "BLUE", "BLUE", "RED"]

# Матрица жолдарының қосындысын тексеру
for s, row in zip(states, T):
    print("T", s, sum(row))
for i, s in enumerate(states):
    print("O", s, emission["RED"][i] + emission["BLUE"][i])
```

## Функциялар: predict() және update()

```python
def predict(b, T):
    """Болжау: b_t^- (j) = Σ b(i) · T[i][j]"""
    return [sum(b[i] * T[i][j] for i in range(3)) for j in range(3)]


def update(pred, likelihood):
    """Жаңарту + нормалау: b_t(j) = P(E|j) · b_t^-(j) / Z"""
    weights = [pred[i] * likelihood[i] for i in range(3)]
    total = sum(weights)
    return [w / total for w in weights]
```

## 4-тапсырма: бір қадам (RED) толық есеп

```python
pred = predict(b0, T)
print("Болжам b1- =", [round(v, 4) for v in pred])
w = [pred[i] * emission["RED"][i] for i in range(3)]
print("Салмақтар w1 =", [round(v, 4) for v in w])
print("Z1 =", round(sum(w), 4))
print("b1 =", [round(v, 4) for v in update(pred, emission["RED"])])
```

## 5-тапсырма: RED, BLUE, BLUE, RED

```python
def run_filter(belief, emission):
    for reading in readings:
        belief = update(predict(belief, T), emission[reading])
        best = states[belief.index(max(belief))]
        print(f"{reading:5} {[round(v, 4) for v in belief]} -> Ең ықтимал: {best}")

run_filter(b0, emission)
```

## 6-тапсырма: эксперимент
Бастапқы үлестірім [0.2, 0.2, 0.6] болғанда:

```python
run_filter([0.2, 0.2, 0.6], emission)
```

Қосымша: RED сигналы әлсіз ажыратқанда [0.6, 0.5, 0.4]:

```python
weak = {"RED": [0.6, 0.5, 0.4], "BLUE": [0.4, 0.5, 0.6]}
run_filter(b0, weak)
```
