```python
values = {'H': 3, 'I': 12, 'J': 8, 'K': 2, 'L': 4, 'M': 6, 'N': 14, 'O': 5}
graph = {'A':['B','C'], 'B':['D','E'], 'C':['F','G'],
         'D':['H','I'], 'E':['J','K'], 'F':['L','M'], 'G':['N','O']}
max_nodes = {'A','D','E','F','G'}

def minimax(node):
    if node in values:
        return values[node]
    child_values = [minimax(c) for c in graph[node]]
    return max(child_values) if node in max_nodes else min(child_values)

print('A =', minimax('A'))
print('B =', minimax('B'), 'C =', minimax('C'))
```

```python
def alpha_beta(node, alpha=float('-inf'), beta=float('inf')):
    if node in values:
        return values[node]
    if node in max_nodes:
        result = float('-inf')
        for child in graph[node]:
            result = max(result, alpha_beta(child, alpha, beta))
            alpha = max(alpha, result)
            if alpha >= beta:
                break
        return result
    else:
        result = float('inf')
        for child in graph[node]:
            result = min(result, alpha_beta(child, alpha, beta))
            beta = min(beta, result)
            if alpha >= beta:
                break
        return result

print('Alpha-Beta A =', alpha_beta('A'))
```

```python
graph = {
    'A': ['B','C'], 'B': ['D','E'], 'C': ['F','G'],
    'D': ['H','I'], 'E': ['J','K'], 'F': ['L','M'], 'G': ['N','O']
}
values = {'H':3, 'I':12, 'J':8, 'K':2, 'L':4, 'M':6, 'N':14, 'O':5}
max_nodes = {'A','D','E','F','G'}

count_minimax = 0
count_ab = 0

def minimax(node):
    global count_minimax
    count_minimax += 1
    if node in values: return values[node]
    vals = [minimax(c) for c in graph[node]]
    return max(vals) if node in max_nodes else min(vals)

def alphabeta(node, alpha=float('-inf'), beta=float('inf')):
    global count_ab
    count_ab += 1
    if node in values: return values[node]
    if node in max_nodes:
        result = float('-inf')
        for c in graph[node]:
            result = max(result, alphabeta(c, alpha, beta))
            alpha = max(alpha, result)
            if alpha >= beta: break
        return result
    result = float('inf')
    for c in graph[node]:
        result = min(result, alphabeta(c, alpha, beta))
        beta = min(beta, result)
        if alpha >= beta: break
    return result

m = minimax('A')
a = alphabeta('A')
print('Minimax:', m, 'түйін саны:', count_minimax)
print('Alpha-Beta:', a, 'түйін саны:', count_ab)
print('Шешім бірдей:', m == a)
```
