```python
# 1. Күйлер, әрекеттер және параметрлер
states = ["A", "B", "C", "D", "E", "S", "F", "H", "I"]
actions = ["UP", "DOWN", "LEFT", "RIGHT"]
gamma = 0.9
theta = 0.001

# Бастапқы мәндер V0(s) = 0
V = {s: 0.0 for s in states}

# Тордың құрылымы (көршілес ұяшықтар)
grid = {
    "A": {"UP": "A", "DOWN": "D", "LEFT": "A", "RIGHT": "B"},
    "B": {"UP": "B", "DOWN": "B", "LEFT": "A", "RIGHT": "C"},
    "C": {"UP": "C", "DOWN": "E", "LEFT": "B", "RIGHT": "G"},  # G - мақсат (+10)
    "D": {"UP": "A", "DOWN": "S", "LEFT": "D", "RIGHT": "D"},  # Оң жағы қабырға (#)
    "E": {"UP": "C", "DOWN": "H", "LEFT": "E", "RIGHT": "X"},  # Сол жағы қабырға (#), оң жағы X (-10)
    "S": {"UP": "D", "DOWN": "S", "LEFT": "S", "RIGHT": "F"},
    "F": {"UP": "F", "DOWN": "F", "LEFT": "S", "RIGHT": "H"},
    "H": {"UP": "E", "DOWN": "H", "LEFT": "F", "RIGHT": "I"},
    "I": {"UP": "X", "DOWN": "I", "LEFT": "H", "RIGHT": "I"},  # Жоғары жағы X (-10)
}

# Сол жаққа және оң жаққа ауытқу бағыттары
ortho_left = {"UP": "LEFT", "DOWN": "RIGHT", "LEFT": "DOWN", "RIGHT": "UP"}
ortho_right = {"UP": "RIGHT", "DOWN": "LEFT", "LEFT": "UP", "RIGHT": "DOWN"}


def reward_of(next_state):
    """Келесі ұяшыққа байланысты марапат."""
    if next_state == "G":
        return 10.0
    if next_state == "X":
        return -10.0
    return -0.1


# 2. Ауысу ықтималдықтары мен марапаттар: transitions(s, a)
def transitions(s, a):
    results = []
    for prob, direction in [(0.8, a), (0.1, ortho_left[a]), (0.1, ortho_right[a])]:
        next_state = grid[s][direction]
        results.append((prob, next_state, reward_of(next_state)))
    return results


# Бір әрекеттің Q-мәні
def q_value(s, a, V):
    q = 0.0
    for prob, next_state, reward in transitions(s, a):
        # Терминалдық күйлердің (G, X) V-мәні 0
        next_val = 0.0 if next_state in ["G", "X"] else V[next_state]
        q += prob * (reward + gamma * next_val)
    return q


# 3. Мәндер итерациясы циклі (Value Iteration)
iteration = 0
while True:
    iteration += 1
    delta = 0.0
    new_V = V.copy()

    for s in states:
        new_V[s] = max(q_value(s, a, V) for a in actions)
        delta = max(delta, abs(new_V[s] - V[s]))

    V = new_V
    if delta < theta:
        break

print("Итерация саны:", iteration)

# 4. Оңтайлы саясат
policy = {}
for s in states:
    best_action = None
    best_q = -float("inf")
    for a in actions:
        q = q_value(s, a, V)
        if q > best_q:
            best_q = q
            best_action = a
    policy[s] = best_action

# 5. Нәтижені шығару
print("\nКүй   V(s)     Ең жақсы әрекет")
for s in states:
    print(f"{s:<5} {V[s]:>7.3f}   {policy[s]}")

# Тор түрінде көрсету
arrows = {"UP": "↑", "DOWN": "↓", "LEFT": "←", "RIGHT": "→"}
layout = [["A", "B", "C", "G"],
          ["D", "#", "E", "X"],
          ["S", "F", "H", "I"]]
print("\nОңтайлы саясат (тор):")
for row in layout:
    line = ""
    for cell in row:
        if cell == "G":
            line += "  +10 "
        elif cell == "X":
            line += "  -10 "
        elif cell == "#":
            line += "   #  "
        else:
            line += f"  {cell}{arrows[policy[cell]]}  "
    print(line)
```
